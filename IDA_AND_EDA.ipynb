Genearte a scatter plot with all the mean columns

```python
cols = ['diagnosis', 'radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean', 'smoothness_mean', 'compactness_mean', 'concavity_mean', 'concave points_mean', 'symmetry_mean', 'fractal_dimension_mean']
sns.pairplot(data=cancer[cols], hue = 'diagnosis', palette = 'rocket')
```

```python
cancer['diagnosis'] = cancer['diagnosis'].map({1:'M', 0:'B'})
```

```python
attribute = "radius_mean"
sns.set_theme(style="whitegrid", palette="muted")
# Draw a categorical scatterplot to show each observation
ax = sns.swarmplot(data=cancer, x=attribute, y="diagnosis", hue="diagnosis")
ax.set(ylabel="")
```

```python
attribute1="radius_mean"
attribute2="radius_worst"
sns.jointplot(data=cancer, x=attribute1, y=attribute2, hue="diagnosis")
```

```python
cols = ['diagnosis', 'radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean', 'smoothness_mean', 'compactness_mean', 'concavity_mean', 'concave points_mean', 'symmetry_mean', 'fractal_dimension_mean']
sns.set_theme(style="ticks")
sns.pairplot(cancer[cols], hue="diagnosis")
```

```python
cols = ['diagnosis', 'radius_mean', 'texture_mean', 'radius_worst', 'texture_worst',]
sns.displot(cancer[cols], kind="kde", rug=True)
```

We chose the heatmap to analyze the correlation among columns and see if the dimension could be reduced.<br>
The swarmplot allows us to clearly separate the scatter plots by diagnosis. Through this, we can see the differences in values based on whether they have an M or B diagnosis. <br>
The jointplot allows us to see the correlation between two columns and also the distributions of those. <br>
This allowed us to see the relationship between two similar variables, for example, radius_mean and radius_worst. We can also see the clear differences depending on the diagnosis. <br>
In the pairplot, we can see that there are clear differences in the distribution of each column depending on the diagnosis. <br>
The displot (kde) makes it possible to see the points that generate the distribution plots.
